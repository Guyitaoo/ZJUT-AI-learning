```python
!pip install jieba
import jieba

text = "我爱自然语言处理，它是一门很有趣的学科。"
# 1. 精确模式 (默认模式)：试图将句子最精确地切开
seg_list_precise = jieba.cut(text, cut_all=False)
print(f"精确模式: {' / '.join(seg_list_precise)}")

# 2. 全模式：把句子中所有可以成词的词语都扫描出来
seg_list_all = jieba.cut(text, cut_all=True)
print(f"全模式: {' / '.join(seg_list_all)}")

# 3. 搜索引擎模式：在精确模式的基础上，对长词再次切分
seg_list_search = jieba.cut_for_search(text)
print(f"搜索引擎模式: {' / '.join(seg_list_search)}")
```

```python
import jieba
def load_stopwords(filepath):
    """
    从指定路径加载停用词表。
    Args:
        filepath (str): 停用词文件的路径
    Returns:
        set: 包含所有停用词的集合 (set)
    """
    print(f"正在从 {filepath} 加载停用词...")
    stopwords = set()
    
    # 1. 打开文件
    #    'r' 表示读取模式
    #    encoding='utf-8' 是处理中文文件的标准做法，防止乱码
    with open(filepath, 'r', encoding='utf-8') as f:
        # 2. 逐行读取
        for line in f:
            # 3. 清理每行：
            #    .strip() 会移除开头和结尾的空白字符（比如换行符 \n 和空格）
            word = line.strip()
            # 4. 确保行不是空的
            if word:
                stopwords.add(word)
    
    print(f"成功加载 {len(stopwords)} 个停用词。")
    return stopwords

# 1. 准备停用词表 (从文件加载)
#    (hit_stopwords.txt
stopwords_filepath = 'hit_stopwords.txt'
stopwords = load_stopwords(stopwords_filepath)

# 打印几个看看
print(f"加载的停用词示例: {list(stopwords)[:5]}")


# 2. 准备原始句子
text = "我爱自然语言处理，它是一门很有趣的学科。"

# 3. 先进行分词
seg_list = jieba.cut(text, cut_all=False)

# 4. 过滤停用词 (这里的逻辑和之前一样)
filtered_words = [word for word in seg_list if word.strip() and word not in stopwords]

print(f"\n原始分词: {' / '.join(jieba.cut(text, cut_all=False))}")
print(f"去除停用词后: {' / '.join(filtered_words)}")
```

```python
# 准备语料 和 分词
!pip install jieba
import jieba

# 1. 原始中文语料
corpus_raw = [
    "我喜欢学习自然语言处理",
    "我讨厌学习线性代数",
    "自然语言处理很有趣，线性代数也很有趣"
]

# 2. 创建一个函数，用于分词并用空格连接
stopwords = {"也", "很", "的", "，"} # 假设一个小的停用词表
def chinese_tokenizer(text):
    words = jieba.cut(text)
    return " ".join([word for word in words if word.strip() and word not in stopwords])
# 3. 处理我们的语料库
corpus_tokenized = [chinese_tokenizer(text) for text in corpus_raw]
print("分词后的语料库:")
for doc in corpus_tokenized:
    print(f"  - {doc}")
```

```python
# 使用 CountVectorizer 实现 TF
from sklearn.feature_extraction.text import CountVectorizer
# 1. 初始化 CountVectorizer
# 它可以看作是"词袋模型"的TF实现
tf_vectorizer = CountVectorizer()
# 2. 拟合和转换我们分好词的语料库
# fit: 学习词典 
# transform: 把句子转换成向量
X_tf = tf_vectorizer.fit_transform(corpus_tokenized)

# 3. 查看结果
print("\n--- TF (CountVectorizer) 结果 ---")
print("词典 (Vocabulary):")
# .get_feature_names_out() 是新版 sklearn 的标准用法
print(tf_vectorizer.get_feature_names_out())

print("\n词频 (TF) 矩阵 (稀疏矩阵表示):")
print(X_tf)

print("\n词频 (TF) 矩阵 (密集数组表示):")
print(X_tf.toarray())
```

```python
!pip install jieba
import jieba

# 1. 原始中文语料
corpus_raw = [
    "我喜欢学习自然语言处理",
    "我讨厌学习线性代数",
    "自然语言处理很有趣，线性代数也很有趣"
]

# 2. 创建一个函数，用于分词并用空格连接
stopwords = {"也", "很", "的", "，"} # 假设一个小的停用词表
def chinese_tokenizer(text):
    words = jieba.cut(text)
    return " ".join([word for word in words if word.strip() and word not in stopwords])
# 3. 处理我们的语料库
corpus_tokenized = [chinese_tokenizer(text) for text in corpus_raw]
print("分词后的语料库:")
for doc in corpus_tokenized:
    print(f"  - {doc}")

    
#  使用 TfidfVectorizer 实现 TF-IDF
from sklearn.feature_extraction.text import TfidfVectorizer

# 1. 初始化 TfidfVectorizer
tfidf_vectorizer = TfidfVectorizer()

# 2. 拟合(fit)和转换(transform)
X_tfidf = tfidf_vectorizer.fit_transform(corpus_tokenized)

# 3. 查看结果
print("\n--- TF-IDF (TfidfVectorizer) 结果 ---")
print("词典 (Vocabulary):")
print(tfidf_vectorizer.get_feature_names_out())

print("\nTF-IDF 权重矩阵 (密集数组表示):")
print(X_tfidf.toarray())

print("\nTF-IDF 权重矩阵 (稀疏矩阵表示):")
print(X_tfidf)
```
