```python
import numpy as np

# 输入数据：类别标签列表
labels = [0, 2, 1, 3, 1, 0]

# 计算类别总数
num_classes = max(labels) + 1  # 4

# 初始化全零矩阵
one_hot_matrix = np.zeros((len(labels), num_classes))

# 填充one-hot编码
for i, label in enumerate(labels):
    one_hot_matrix[i, label] = 1

print(one_hot_matrix)
```

```python
import numpy as np

# 输入数据：文本句子列表
sentences = [
    "我 毕业 于 温州第二高级中学",
    "现 就读 于 浙江工业大学"
]

# 步骤1: 构建词汇表
token_index = {}
for sentence in sentences:
    for word in sentence.split():
        if word not in token_index:
            token_index[word] = len(token_index)  # 索引从0开始

# 步骤2: 创建编码矩阵
results = np.zeros((len(sentences), len(token_index)))

# 步骤3: 填充编码矩阵
for i, sentence in enumerate(sentences):
    for word in sentence.split():
        j = token_index[word]
        results[i, j] = 1

print("词汇表:", token_index)
print("编码矩阵:\n", results)
```
