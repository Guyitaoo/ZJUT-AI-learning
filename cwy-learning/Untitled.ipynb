原理，监督学习是在通过一系列拥有结果的数据自行建立起一个类似函数的东西，从而预测下一次输入输出的结果
举个例子，电脑程序中我们可能碰见过简单的编程题目，比如给你一个公式：房价 = 房子面积*A + C，输入房子面积让我们求房价，在这里我们是一个知道过程求结果的过程，这里的函数 ：房价 = 房子面积*A + C 对我们来说是已知的，但是实际情况中，我们可能并不知道这个过程或者函数或者说房价和房子面积的关系，这个函数可能是一条曲线，一个圆，一个指数方程，甚至一个三角函数，我们不知道，但是我们可以收集一系列数据，让机器去拟合出这个关系，这个就是监督学习中最基础的线性回归
ps：监督学习和非监督学习最本质的区别就是有无结果反馈（例子中的房价）

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. 构造模拟数据（广告投入x，销售额y）
x = np.array([5, 10, 15, 20, 25, 30])  # 自变量
y = np.array([10, 20, 25, 40, 45, 55]) # 因变量
m = len(x) # 样本数

# 2. 计算均值
x_mean = np.mean(x)
y_mean = np.mean(y)

# 3. 用最小二乘法公式计算w和b
numerator = np.sum((x - x_mean) * (y - y_mean)) # 分子
denominator = np.sum((x - x_mean) ** 2)        # 分母
w = numerator / denominator
b = y_mean - w * x_mean

# 4. 生成预测值
y_pred = w * x + b

# 5. 可视化：散点图+拟合直线
plt.scatter(x, y, color='blue', label='真实数据')
plt.plot(x, y_pred, color='red', label=f'拟合直线: y={w:.2f}x+{b:.2f}')
plt.xlabel('广告投入')
plt.ylabel('销售额')
plt.legend()
plt.show()

# 6. 计算评估指标MSE和R²
mse = np.mean((y_pred - y) ** 2)
r2 = 1 - (np.sum((y_pred - y) ** 2) / np.sum((y - y_mean) ** 2))
print(f"MSE: {mse:.2f}")
print(f"R²: {r2:.2f}")
```

这里可以看看豆包的代码，一个非常标准的手动构建监督学习的过程
搭建原理：上述代码中通过最小二乘法一次性得到最终结果不存在迭代等使一次函数逐渐贴近目标值的过程
