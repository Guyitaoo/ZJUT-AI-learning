### 数据读取

- 以旅游景点游客评论为例
- python 读取excel文件, 按sheet读取数据

### 数据概况

- 读取sheet中的数据概况信息, 并打印输出
- 景点信息、评论数目、景点详情
- 数据来源

### 数据清洗

 - 读取景点评论数据sheet
 - 数据去重: 用户名、用户ID、评论内容

### 数据统计

- 对景点评论数据的各个特征进行统计分析，并可视化（直方图、饼图等）
- 评论数目、用户IP属地、评论日期、评分、旅游类型等

### 评论内容

- 读取数据表中的评论内容 (content)
- 合并多个网站同一景点的评论内容, 保存为txt文件

### 文本预处理

- 统一全半角、标点、英文大小写、去除表情符号......

### jieba分词

- jieba分词、词性标注、去停用词
- 自定义词典

### 构建词库

- 词频统计
- 按词性进行词频统计
- 可视化：选取高频词生成词云图（可按词性区分）

### One-Hot表示

- 生成词的one-hot表示

### TF、TF-IDF

- 计算词项的词频TF
- 计算词项的TF-IDF

### 特征词选择

- 根据词频TF或TF-IDF进行特征词选择

### 词袋模型（文本表示）

- 基于选择的特征词, 通过以TF或TF-IDF为权重，对特征词的one-hot向量进行加权, 生成每条评论的文本向量

### 相似度计算

- 基于评论的文本向量, 计算评论之间的相似度
